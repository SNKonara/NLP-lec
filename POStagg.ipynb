<h1> POS tagging </h1>

```python
import nltk
from nltk import pos_tag, word_tokenize
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# download nltk resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt_tab')
```

```python
text= "I love python programming and I enjoy coding in Python"
```

```python
tokens =  word_tokenize(text)
```

```python
tags = pos_tag(tokens)
print(tags)
```

```python
# extract only tags

pos_counts ={}
for word, tag in tags:
    if tag in pos_counts:
        pos_counts[tag] += 1
    else:
        pos_counts[tag] = 1
```

```python
pos_counts
```

```python
#...............Visualization ......................
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=list(pos_counts.keys()), y=list(pos_counts.values()))
plt.title("POS Tag Counts")
plt.xlabel("POS Tag")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(7,7))
plt.pie(pos_counts.values(), labels=pos_counts.keys(), autopct='%1.1f%%',startangle=140)
plt.title("POS Tag Distribution")
plt.show()
```
